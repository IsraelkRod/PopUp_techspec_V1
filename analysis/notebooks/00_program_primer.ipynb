# 00 · SBIR/STTR Program Primer
### The foundation for the Texas SBIR/STTR Impact Analysis

This notebook explains **how the SBIR/STTR program actually works** — Phases 0 through III —
because the phase model is the analytical backbone of everything that follows (the
commercialization funnel, the "are companies stuck cycling awards?" question, survival, all
derive from it). No award data is required to run this notebook.

> **SBIR** = Small Business Innovation Research. **STTR** = Small Business Technology Transfer
> (same phase structure, but STTR *requires* partnering with a research institution, e.g. a
> university, which must perform ≥ 30% of the work).

## The phases at a glance

| Phase | Purpose | Typical award size | Duration | The key rule |
|---|---|---|---|---|
| **Phase 0** | *Not a federal phase.* State / local + **SBDC** help (mentoring, proposal review, sometimes matching funds) to get a company **ready to win Phase I**. This is where the Small Business Development Center story begins. | State-dependent | — | Positions a firm to compete; not an award itself |
| **Phase I** | **Feasibility / proof of concept** — can the idea work? | ~$150K–$323K (SBIR); STTR ~$150K | 6–12 months | Establishes technical merit **and** commercial potential |
| **Phase II** | **R&D build-out / prototype** — develop the thing | ~$1M–$2.15M (SBIR); STTR ~$1M | ~24 months | **You must generally win Phase I first** (exception: "Direct-to-Phase-II") |
| **Phase III** | **Commercialization** — bring it to market | **No SBIR/STTR dollars** | No time limit | Funded by sales, private capital, or **sole-source follow-on federal contracts** |

*Dollar figures are current SBA guideline caps (2025–26). Individual agencies often set lower
internal limits (e.g. DOE Phase I ~$200K, USDA Phase I ~$175K, NSF Phase II ~$1.25M).*

```python
# A self-contained timeline of the phases (no data needed).
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

phases = [
    ('Phase 0',  'SBDC / state prep',      '(no federal $)',   '#9aa0a6'),
    ('Phase I',  'Feasibility',            '~$150K–$323K, 6–12 mo', '#4285f4'),
    ('Phase II', 'R&D / prototype',        '~$1M–$2.15M, ~24 mo',  '#1a73e8'),
    ('Phase III','Commercialization',      'no SBIR $, no limit',   '#34a853'),
]
fig, ax = plt.subplots(figsize=(11, 2.6))
x = 0
for name, sub, meta, color in phases:
    w = 2.6
    ax.add_patch(FancyBboxPatch((x, 0), w, 1, boxstyle='round,pad=0.02,rounding_size=0.08',
                                fc=color, ec='white', lw=2))
    ax.text(x + w/2, 0.62, name, ha='center', va='center', color='white', fontsize=12, fontweight='bold')
    ax.text(x + w/2, 0.32, sub,  ha='center', va='center', color='white', fontsize=9)
    ax.text(x + w/2, -0.35, meta, ha='center', va='center', color='#444', fontsize=8)
    if x > 0:
        ax.annotate('', xy=(x-0.02, 0.5), xytext=(x-0.38, 0.5),
                    arrowprops=dict(arrowstyle='->', color='#888', lw=1.5))
    x += w + 0.4
ax.set_xlim(-0.5, x); ax.set_ylim(-0.7, 1.2); ax.axis('off')
ax.set_title('SBIR/STTR phase progression', fontsize=13, fontweight='bold', loc='left')
plt.tight_layout(); plt.show()
```

## Your questions, answered directly

**“Does winning Award 1 (Phase I) mean I *don't* get to go to Award 2 (Phase II)?”**
The opposite. Phase I is normally the **prerequisite** for Phase II — you generally can't get a
Phase II without first having a Phase I on that technology. **But** winning Phase I does **not
guarantee** Phase II: it is re-competed, and only a fraction of Phase I winners are selected for
Phase II. That fraction — the **Phase I → Phase II transition rate** — is one of our headline
impact metrics (Module E). (Exception: some agencies allow **Direct-to-Phase-II** if feasibility
was already proven elsewhere.)

**“Does Award 2 (Phase II) mean I'm going to commercialize?”**
No, not automatically. Phase II is where you *build*. **Phase III (commercialization) is a
separate step and is not guaranteed** — many awardees never reach it. Phase III uses **no SBIR
money**; it is funded by real customers, private investment, or non-SBIR follow-on federal
contracts (which agencies can award sole-source, without re-competing).

**“Should a company always stay in Phase II territory? Is the purpose just to keep getting
awards and never commercialize?”**
That pattern has a name: the **“SBIR mill”** — firms that keep cycling Phase I/II awards as a
revenue stream without ever commercializing. The *program's stated intent* is the opposite:
Phase III commercialization is the goal. Our analysis **measures both** — who progresses toward
commercialization vs. who cycles awards (Module E, the “cycler vs. commercializer” segmentation).

**“How does a company keep getting awards?”**
By running **multiple projects in parallel**: new topics each solicitation cycle, and separate
Phase I → Phase II sequences on *different* technologies. A single firm can hold many awards
across agencies at once — which is why we deduplicate to **unique firms** when counting companies.

## What it takes to *get* an award, and what happens *after*

**To get one (Phase I):** you must be a U.S. small business (< 500 employees, > 50% U.S.-owned),
respond to a specific agency **topic/solicitation**, and win a competitive technical + commercial
review. The Principal Investigator must be primarily employed by the company. (STTR additionally
requires a research-institution partner doing ≥ 30% of the work.)

**After you get it:** you deliver against the funded scope on a fixed timeline (Phase I: 6–12
months; Phase II: ~24 months), report results, and — if you want to continue — compete for the
next phase. Awards are grants *or* contracts depending on the agency (e.g. DoD typically uses
contracts; NIH/NSF typically grants), which changes reporting and IP/data-rights obligations.

**What an award “means” and how long it lasts:** it is a fixed-scope, fixed-term R&D commitment,
not open-ended funding. It ends when the period of performance ends (`contract_end_date` in the
data). Continuing means winning the *next* competition — there is no automatic renewal.

## SBIR vs STTR — quick contrast

| | SBIR | STTR |
|---|---|---|
| Who does the work | The small business does the majority | Small business **+ research institution**; RI does ≥ 30% |
| PI employment | PI primarily employed by the company | PI may be at the company **or** the research institution |
| Intent | Small-business-led innovation | Bridge university research → market |

Both share the same Phase I / II / III structure and similar dollar bands.
